```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY") or os.getenv("GEMINI_API_KEY")

# List all models available for your API key
client = genai.Client(api_key=api_key)

print("=== Available Google Models ===")
for model in client.models.list():
    # Filter for models that support generateContent
    actions = getattr(model, 'supported_actions', []) or getattr(model, 'supported_generation_methods', [])
    print(f"• {model.name}")
```

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY") or os.getenv("GEMINI_API_KEY")

# Recommended models:
# - "gemini-2.5-flash" (or "gemini-2.0-flash" / "gemini-1.5-flash")
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0.7,
    api_key=api_key,
)

# Simple text prompt test
response = llm.invoke("Explain Signed Distance Functions (SDF) in two sentences.")
print("=== Text Response ===")
print(response.content)
```

```python
import base64
from pathlib import Path
from langchain_core.messages import HumanMessage

# Multimodal image test (testing with render artifact if available)
image_path = Path(r"../eval_artifacts/bench_01_honeycomb_house_render.png")

if image_path.exists():
    with open(image_path, "rb") as img_file:
        img_b64 = base64.b64encode(img_file.read()).decode("utf-8")

    message = HumanMessage(
        content=[
            {"type": "text", "text": "Describe what 3D geometric shape or object is visible in this render."},
            {
                "type": "image_url",
                "image_url": {"url": f"data:image/png;base64,{img_b64}"},
            },
        ]
    )
    mm_response = llm.invoke([message])
    print("=== Multimodal Image Response ===")
    print(mm_response.content)
else:
    print(f"Sample image not found at {image_path}, skipping multimodal test.")
```
